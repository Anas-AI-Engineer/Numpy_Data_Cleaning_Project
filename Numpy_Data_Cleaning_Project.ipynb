```python
import numpy as np

#1 dataset generator
np.set_printoptions(precision=2, suppress=True)

np.random.seed(42)

data = np.column_stack((
   np.random.randint(10, 20, 100),
   np.random.randint(20, 30, 100),
   np.random.randint(30, 40, 100),
   np.random.randint(40, 50, 100)
))

print("Orignal Data:", data[:10])

print("Data shape:", np.shape(data))

#missing Values
data = data.astype(float)

missing_values = np.random.choice(
    data.size, 10, replace=False
)

data.flat[missing_values] = np.nan

print("Missing Values:")
print(np.isnan(data).sum())

#3 missing values imputation

def impute_missing(data, method= "mean"):

    cleaned = data.copy()

    for col in range(cleaned.shape[1]):

        columns = cleaned[:,col]

    if method == "mean":
        value = np.nanmean(columns)

    elif method == "median":
        value = np.nanmedian(columns)

    else:
        raise ValueError("Invalid Method")

    cleaned[np.isnan(cleaned[:,col]), col] = value

    return cleaned

cleaned_data = impute_missing(data, method="median")

print("After missing value imputation")
print(np.isnan(cleaned_data).sum())

#4 remove duplicate
def remove_duplicates(data):
    
    unique_data = np.unique(data, axis=0)

    return unique_data

cleaned_data = remove_duplicates(cleaned_data)    

print("After Missing Duplicates:")
print("Final data shape:", cleaned_data.shape)

print("Final Dataset:", cleaned_data[:10])

print("Final Missing Values:", np.isnan(cleaned_data).sum())
    
    

```
